# House Price Predictor — Multiple Linear Regression
Run from the project root or adjust the project path in the first cell. This notebook generates exploration and diagnostic visualizations. Historical prices, not current valuations.

```
from pathlib import Path
import sys
import pandas as pd
import matplotlib.pyplot as plt
project = Path.cwd()
if not (project / "train_model.py").exists():
    project = project.parent
sys.path.insert(0, str(project))
from train_model import train, load_data
raw, data = load_data()
display(raw.head())
display(raw.describe())
display(raw.isna().sum())
print("Raw rows:", len(raw), "Usable rows:", len(data))
```

```
plt.figure(figsize=(9,5))
plt.scatter(data.sqft_living, data.price, alpha=0.4)
plt.xlabel("Living area (sq ft)")
plt.ylabel("Price (dataset units)")
plt.title("Living area vs. historical price")
plt.show()
```

```
model, report, comparison, data, training = train()
display(pd.DataFrame({"MLR": report["model"], "Baseline": report["baseline"]}).T)
display(comparison.head(20))
```

```
fig, axes = plt.subplots(1,2,figsize=(13,5))
axes[0].scatter(comparison.Actual, comparison.Predicted, alpha=0.4)
lo = min(comparison.Actual.min(), comparison.Predicted.min())
hi = max(comparison.Actual.max(), comparison.Predicted.max())
axes[0].plot([lo,hi],[lo,hi],"r--")
axes[0].set(xlabel="Actual price", ylabel="Predicted price", title="Actual vs. predicted")
axes[1].scatter(comparison.Predicted, comparison.Residual, alpha=0.4)
axes[1].axhline(0, color="red", linestyle="--")
axes[1].set(xlabel="Predicted price", ylabel="Residual", title="Residual diagnostics")
plt.tight_layout()
plt.show()
```

## Limitations
MAE and RMSE are in dataset price units. R² is not an accuracy percentage. Prices date to 2014. Extreme positive prices were retained. Rare unseen test cities use the reference-city encoding. Test performance does not establish future-market accuracy.
